```python

def fractional_knapsack(values, weights, capacity):
    n = len(values)

    items = []
    for i in range(n):
        ratio = values[i] / weights[i]
        items.append((ratio, values[i], weights[i]))

    items.sort(reverse=True)

    total_value = 0

    for ratio, value, weight in items:
        if capacity == 0:
            break

        if weight <= capacity:
            total_value += value
            capacity -= weight
        else:
            total_value += ratio * capacity
            capacity = 0

    return total_value


values = [60, 100, 120]
weights = [10, 20, 30]
capacity = 50

print("Maximum value:", fractional_knapsack(values, weights, capacity))
```
